# Normative Baseline Model & Inference Pipeline Verification


```python
import os
import joblib
import numpy as np
import pandas as pd

model_path = os.path.join('models', 'healthy_normative_baseline.joblib')
if os.path.exists(model_path):
    bundle = joblib.load(model_path)
    pipeline = bundle['pipeline']
    classes = bundle['target_classes']
    feature_names = bundle['feature_names']
    print('Baseline model loaded successfully.')
    print('Classes:', classes)
    print('Features:', feature_names)
```

```python
test_feat_path = os.path.join('data', 'test_adv_advanced_features.csv')
if os.path.exists(test_feat_path):
    df_test = pd.read_csv(test_feat_path)
    sample_row = df_test[feature_names].iloc[[0]]
    probs = pipeline.predict_proba(sample_row)[0]
    pred_class = classes[np.argmax(probs)]
    confidence = np.max(probs)
    print(f'Predicted Movement: {pred_class} ({confidence*100:.1f}% confidence)')
```
